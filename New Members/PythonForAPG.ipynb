### Integers and Floats

Python has two main data types for storing numbers:

- **Integer (`int`)**: A whole number with no decimal point, such as `5`, `20`, or `-10`.
- **Float (`float`)**: A number containing a decimal point, such as `5.2`, `20.0`, or `-10.5`.

Both integers and floats can be used in mathematical operations.

When an integer and a float are used together in a calculation, the result will generally be a float.

### Type Conversion

We can convert between numerical data types using:

- `int()` to convert a value to an integer.
- `float()` to convert a value to a float.

**Note:** Converting a float to an integer does not round the number. It removes the decimal part.

If we want to round a number instead, we can use `round()`.

```python
# Integers and floats
numberOfSimulations = 20
velocity = 25.5

print("number of simulations ",numberOfSimulations)
print(type(numberOfSimulations))
print()


print(velocity)
print(type(velocity))
print()


# Using ints and floats together
result = numberOfSimulations + velocity

print("result ", result)
print("result type: ", type(result))
print()


# Type conversion
integerVelocity = int(velocity)
floatSimulations = float(numberOfSimulations)

print("int velocity : ", integerVelocity)
print("float simulations, ",floatSimulations)
print()


# int() vs round()
number = 5.7

print("turning number into int :", int(number))
print("vs")
print("using rounding function", round(number))
```

### Mathematical Operators

Python can perform mathematical calculations using **operators**.

The main mathematical operators are:

- `+` Addition
- `-` Subtraction
- `*` Multiplication
- `/` Division
- `**` Exponentiation

We can also update the value stored in an existing variable.

Python provides shorthand operators for doing this:

### If, Elif and Else Statements

So far, our code has run from top to bottom and every line has been executed.

Conditional statements allow our program to make **decisions**. We can tell Python to only run certain code when a particular condition is met.

### `if`

An `if` statement checks whether a condition is `True`.

If the condition is `True`, the indented code underneath it is executed. If the condition is `False`, Python skips that code.

### `else`

An `else` statement can be used to specify what should happen when the `if` condition is `False`.

### `elif`

`elif`, meaning **else if**, allows us to check additional conditions.

Python checks the conditions from top to bottom and executes the first one that is `True`.
you can have multiple elif statements to check for many conditions, they are checked in the order you write them, so if the first is true its code is executed and the rest ignored, even if one later down is also true.

Indentation is very important in Python.

### Multiple Conditions

Sometimes we want to check more than one condition at once. We can use:

- `and` when **both** conditions must be `True`
- `or` when **at least one** condition must be `True`
- `not` to reverse a Boolean condition

```python
downforce = 450
targetDownforce = 400

# Basic if statement
if downforce > targetDownforce:
    print("Downforce target achieved")

print()


# If and else
if downforce > targetDownforce:
    print("Downforce target achieved")
else:
    print("Downforce target not achieved")

print()


# If, elif and else
if downforce > 500:
    print("High downforce")

elif downforce > 400:
    print("Medium downforce")

elif downforce > 450:
    # TODO tell me why this elif block will not run?
    print("medium high downforce")

else:
    print("Low downforce")

print()


# Multiple conditions using and
downforce = 450
drag = 180

if downforce > 400 and drag < 200:
    print("Aero targets achieved")

print()


# Multiple conditions using or
if downforce < 300 or drag > 250:
    print("Aero performance needs improvement")
else:
    print("Aero performance is acceptable")

print()


# Using not
simulationFailed = False

if not simulationFailed:
    print("Simulation completed successfully")
else:
    print("Simulation failed")
```

### For Loops
> this section is very important for the work we will be doing in apg for cfd sweeps and optimisation

A **loop** allows us to repeat a section of code multiple times without having to write the same code repeatedly.

A `for` loop is used when we want to repeat something a known number of times or work through a collection of values.

### `range()`

The `range()` function generates a sequence of integers that a `for` loop can work through.

For example, `range(5)` generates numbers from `0` to `4`.

Remember that Python starts counting from `0`, and the final value in `range()` is **not included**.

We can also specify a starting value and step size:

- `range(5)` → 0, 1, 2, 3, 4
- `range(2, 6)` → 2, 3, 4, 5
- `range(0, 10, 2)` → 0, 2, 4, 6, 8

Just like with `if` statements, indentation tells Python which code belongs inside the loop.

The variable used in the `for` statement changes value each time the loop repeats. Each repetition of a loop is called an **iteration**.

you can also for loop through a list, a list is just a collection of things, and is a new data type!

```python
# Basic for loop
for i in range(5):
    print(i)

print()


# Choosing a start and end
for i in range(2, 6):
    print(i)

print()


# Choosing a step size
for i in range(0, 10, 2):
    print(i)

print()

# NEW DATA TYPE: LIST
# lists can hold any data type within them
ourList = [0, 1, 2, 3, 4, 5, "exampleString", True, False]
for item in ourList:
    print(item)

# Using the loop variable in a calculation
for velocity in range(10, 31, 5):
    velocitySquared = velocity**2

    print(f"Velocity: {velocity} m/s")
    print(f"Velocity Squared: {velocitySquared} m²/s²")

print()


# Repeating an aerodynamic calculation
# todo, we will be doing this as part of the real apg work, but . . .
# todo . . . in a more complex system using starccm and solidworks
airDensity = 1.225
referenceArea = 1.5
liftCoefficient = -2.4

for velocity in range(10, 31, 5):
    downforce = -0.5 * airDensity * velocity**2 * referenceArea * liftCoefficient

    print(f"At {velocity} m/s, Downforce = {round(downforce, 2)} N")
```

### While Loops

A `while` loop repeatedly runs a section of code **while a condition remains `True`**.

Unlike a `for` loop, a `while` loop does not necessarily run a known number of times. Instead, it continues until its condition becomes `False`.

This makes `while` loops useful when we want something to continue until a particular condition is reached.

The condition is checked before every iteration of the loop.

### Updating the Condition

It is important that something inside the loop eventually causes the condition to become `False`.

If the condition never becomes `False`, the loop will continue forever. This is called an **infinite loop**. if u get an infinite loop u will need to restart your programming softward lol

### For Loop vs While Loop

A `for` loop is generally used when we know what values we want to loop through or how many iterations we want.

A `while` loop is generally used when we want to continue until a particular condition is met.

```python
# Basic while loop
iteration = 0

while iteration < 5:
    print(iteration)
    iteration += 1

print()


# Continue until a target is reached
downforce = 200
targetDownforce = 400

while downforce < targetDownforce:
    print(f"Current Downforce: {downforce} N")
    downforce += 50

print(f"Target reached: {downforce} N")

print()


# Simulation example
iteration = 0
maximumIterations = 5
simulationConverged = False

while iteration < maximumIterations and not simulationConverged:
    iteration += 1

    print(f"Running iteration {iteration}")

    if iteration == 4:
        simulationConverged = True

print()

print(f"Simulation Converged: {simulationConverged}")
```

### Lists

I mentioned these earlier in the for loops section, a **list** allows us to store multiple values inside a single variable.

Lists are created using square brackets `[ ]`, with each item separated by a comma.

A list can contain numbers, strings, Booleans, or even a mixture of different data types.

### Indexing

Each item in a list has an **index**, which represents its position.

Python starts counting from `0`, so the first item has an index of `0`, the second has an index of `1`, and so on.

Negative indices can be used to count backwards from the end of a list. An index of `-1` refers to the final item.

### Adding and Removing Items

Python provides several methods for modifying lists.

- `append()` adds an item to the end of a list.
- `insert()` adds an item at a specified index.
- `remove()` removes a specified item.
- `pop()` removes an item at a specified index.

The `len()` function can be used to find the number of items in a list.

### Looping Through Lists

A `for` loop can be used to work through every item in a list.

This is extremely useful when we want to perform the same calculation or operation on multiple values.

```python
# Creating lists
velocities = [10, 15, 20, 25, 30]

components = ["Front Wing", "Floor", "Rear Wing"]

print(velocities)
print(components)

print()


# Accessing items using their index
# remember, 0 is the first term, this is a very common mistake
print(velocities[0])
print(velocities[2])
print(velocities[-1])

print()


# Changing an item
velocities[1] = 17

print(velocities)

print()


# Adding items
velocities.append(35)
velocities.insert(0, 5)

print(velocities)

print()


# Removing items
velocities.remove(20)

print(velocities)

velocities.pop(0)

print(velocities)

print()


# Finding the length of a list
print(len(velocities))

print()


# Looping through a list
velocities = [10, 15, 20, 25, 30]

for velocity in velocities:
    print(f"Velocity: {velocity} m/s")

print()


# Using a list for an aerodynamic calculation
airDensity = 1.225
referenceArea = 1.5
liftCoefficient = -2.4

# this is a velocity sweep, lets us test various conditions for our solver
for velocity in velocities:
    downforce = -0.5 * airDensity * velocity**2 * referenceArea * liftCoefficient

    print(f"At {velocity} m/s, Downforce = {round(downforce, 2)} N")
```

### Functions

A **function** is a reusable block of code designed to perform a specific task.

Functions allow us to write a piece of code once and then use it multiple times throughout a program. This helps make code shorter, easier to understand, and easier to modify.

Functions are created using the `def` keyword followed by the function name and brackets.

### Parameters and Arguments

Functions can be given information to work with.

A **parameter** is a variable defined when the function is created.

An **argument** is the actual value given to the function when it is called.

A function can have multiple parameters, separated by commas.

### Default Parameters

A parameter can be given a **default value** when a function is defined. If no argument is provided when the function is called, Python will use the default value instead.

### Returning Values

A function can use `return` to send a result back to the part of the program that called it.

The returned value can then be stored in a variable, printed, or used in another calculation.

Functions are particularly useful for engineering calculations because the same calculation can be reused with many different inputs.

### Docstrings
A **docstring** is a short description placed inside a function to explain what the function does, its inputs, and what it returns. you should get in the habbit of writing these, when code gets big its very easy to forget what it does



```python
# Function with a parameter
def PrintVelocity(velocity):
    print(f"Velocity: {velocity} m/s")

PrintVelocity(20)
PrintVelocity(25)

print()


# Function with multiple parameters and a return value
def CalculateDownforce(airDensity, velocity, liftCoefficient, referenceArea = 1):
    downforce = -0.5 * airDensity * velocity**2 * referenceArea * liftCoefficient

    return downforce

# notice, only 3 arguements were give, because reference area will default to 1
downforce = CalculateDownforce(1.225, 20, -2.4)

# notice, we can also change the value of reference area if we want to
# it just defaults to 1 if we say nothing

# you should write referenceArea= the value you want, so its more clear when you have . . .
# . . . a case with multiple default arguments
downforce = CalculateDownforce(1.225, 20, -2.4, referenceArea= 2)



print(f"Downforce: {round(downforce, 2)} N")

print()


# Reusing a function with a loop
velocities = [10, 15, 20, 25, 30]

for velocity in velocities:
    downforce = CalculateDownforce(1.225, velocity, 1.5, -2.4)

    print(f"At {velocity} m/s, Downforce = {round(downforce, 2)} N")
```

### Classes

A **class** is a way of grouping related data and functions together into a single structure.

An **object** is an individual instance of a class. For example, we could create an `AeroSimulation` class and then create different simulation objects from it.

The `__init__()` function runs automatically when a new object is created and is used to set its initial values.

`self` refers to the specific object being used.

Variables stored inside an object are called **attributes**, while functions belonging to a class are called **methods**.

Classes become particularly useful as programs get larger, and we want to keep related data and functionality organised together.

> Classes also get very complex and have alot of compsci theory behind them since they are incredibly flexible, so this is just an intro so u can see what they are about when u see someone else's code.

```python
class AeroSimulation:

    def __init__(self, velocity, liftCoefficient, dragCoefficient):
        self.velocity = velocity
        self.liftCoefficient = liftCoefficient
        self.dragCoefficient = dragCoefficient

    def PrintResults(self):
        print(f"Velocity: {self.velocity} m/s")
        print(f"Lift Coefficient: {self.liftCoefficient}")
        print(f"Drag Coefficient: {self.dragCoefficient}")


# Create an AeroSimulation object
ourSimulation = AeroSimulation(20, -2.4, 0.9)

# Access its attributes
print(ourSimulation.velocity)
print(ourSimulation.liftCoefficient)
print(ourSimulation.dragCoefficient)

print()


# Call a method belonging to the object
ourSimulation.PrintResults()
```

```python
# an import is normally done at the very top of our file, or in jupyter the first cell,
# here it is in this cell to show u how to import
import numpy as np


# Normal Python list
velocitiesList = [10, 15, 20, 25, 30]
print(velocitiesList)

print()


# NumPy array
velocitiesArray = np.array([10, 15, 20, 25, 30])
print(velocitiesArray)

print()


# Multiplying a normal list
print(velocitiesList * 2)

# Multiplying a NumPy array
print(velocitiesArray * 2)
print("notice the difference in the above two outputs ^")
print()


# Mathematical operations can be applied to the entire array
print(velocitiesArray + 5)
print(velocitiesArray - 5)
print(velocitiesArray ** 2)
print(velocitiesArray / 2)
```

### Creating NumPy Arrays

There are several ways to create NumPy arrays depending on what values we need.

`np.array()` converts an existing list of values into a NumPy array.

`np.arange()` creates an array of evenly spaced values using a **start, stop and step size**. Like `range()`, the stop value is not included.

`np.linspace()` creates an array between a start and end value, but instead of specifying the step size, we specify the **number of values** we want.

`np.linspace()` is particularly useful in engineering when we want a specific number of evenly spaced points between two values.

```python
# No import cos we already did that last cell

# Creating an array from a list
velocities = np.array([10, 15, 20, 25, 30])

print(velocities)

print()


# np.arange(start, stop, step)
velocities = np.arange(10, 31, 5)

print(velocities)

print()


# np.linspace(start, stop, numberOfValues)
velocities = np.linspace(10, 30, 5)

print(velocities)

print()


# Creating more closely spaced points
yawAngles = np.linspace(-10, 10, 9)

print(yawAngles)
```
